```python
import numpy as np
import pandas as pd
df = pd.read_csv(r'D:\git.ir\Daneshjooyar-Data-Science-and-ML\03-Pandas Basics to Advance\tips.csv')
df.head()
```

```python
def last_four(num):
    return str(num)[-4:]
```

```python
df['CC Number'][0]
```

```python
last_four(df['CC Number'][233])
```

```python
df['last four'] = df['CC Number'].apply(last_four)
df['last four']
```

```python
df.head()
```

```python
df['total_bill'].mean()
```

```python
def yelp(price):
    if price < 10:
        return '$'
    elif price >= 10 and price < 30:
        return '$$'
    else:
        return '$$$'
```

```python
df['Expensive'] = df['total_bill'].apply(yelp)
df.head(23)
```

```python
lambda num: num**2
```

```python
df['total_bill'].apply(lambda bill: bill * 0.18 )
```

```python
def quality(total_bill, tip):
    if tip / total_bill > 0.20:
     return 'Generous'
    else:
        return 'Other'

df['quality'] = df.apply(lambda row: quality(row['total_bill'], row['tip']), axis=1)
df.head()
    
```

```python
df['Tip Quality'] = np.vectorize(quality)(df['total_bill'], df['tip'])
df.head(7)
```
