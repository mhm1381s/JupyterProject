````python
# pd.Panel was removed in pandas 0.25.0.
# Use a dict of DataFrames combined with pd.concat instead.
import pandas as pd
import numpy as np

data = {'Item1': pd.DataFrame(np.random.randn(4, 3)),
        'Item2': pd.DataFrame(np.random.randn(4, 2))}

# Combine the items along the columns -> MultiIndex columns (item, column)
p = pd.concat(data, axis=1)

# Equivalent of the old Panel.minor_xs(1): cross-section on the minor axis
print(p.xs(1, axis=1, level=1))
````
