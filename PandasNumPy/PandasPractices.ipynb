```python
hotels['total_stays_days'] = hotels['stays_in_week_nights'] + hotels['stays_in_weekend_nights']

round(hotels['total_stays_days'].mean(), 2)
```

```python
hotels['total_paid'] = hotels['total_stays_days'] * hotels['adr']
hotels['total_paid']
```

```python
# فرض می‌کنیم اسم دیتافریم hotels است و ستون ایمیل هم email نام دارد
result = hotels[hotels['total_of_special_requests'] == 5][['name', 'email']]
print(result)
```

```python
def nameOfRequestEqualFive(df):
    # فیلتر کردن کسانی که دقیقا ۵ درخواست ویژه دارند
    filtered_df = df[df['total_of_special_requests'] == 5]
    # برگرداندن نام و ایمیل آن‌ها
    return filtered_df[['name', 'email']]

# نحوه صدا زدن تابع:
print(nameOfRequestEqualFive(hotels))
```
