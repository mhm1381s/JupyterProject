```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define x values
x = np.linspace(-10, 10, 500)

# Create figure and axes
fig, ax = plt.subplots(figsize=(9, 6))

# Plot functions
ax.plot(x, x**2, label='x^2', linewidth=2)
ax.plot(x, x**3, label='x^3', linewidth=2)

# Set axis limits
ax.set_xlim(-10, 10)
ax.set_ylim(-20, 20)

# Draw coordinate axes
ax.axhline(0, color='black', linewidth=1.2)
ax.axvline(0, color='black', linewidth=1.2)

# Add grid
ax.grid(True, linestyle='--', alpha=0.5)

# Add labels
ax.set_xlabel('X')
ax.set_ylabel('Y')

# Add title
ax.set_title('Graphs of Y = X^2 and Y = X^3')

# Show legend
ax.legend()

# Display graph
plt.show()
```
