```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.linspace(0,10,11)
b = a ** 4
```

```python
x = np.arange(0,10)
```

```python
y = 2 * x
y
```

```python
x
```

```python
fig , axes = plt.subplots()
axes.plot(x, y, 'r') # r = red color 
axes.set_xlabel('X')
axes.set_ylabel('Y')
axes.set_title('title')

plt.show()
```

```python
fig , axes = plt.subplots()
axes.plot(x, y, 'b') # b = blue color
axes.set_xlabel('X')
axes.set_ylabel('Y')
axes.set_title('title')

plt.show()
```

```python
fig , axes = plt.subplots()
axes.plot(x, y, 'g') # g = Green color
axes.set_xlabel('X')
axes.set_ylabel('Y')
axes.set_title('title')

plt.show()
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 2)
axes[0].plot(a, b)
axes[1].plot(x, y)
```

```python
fig, axes = plt.subplots(nrows = 2, ncols= 2)
axes[0][0].plot(a, b)
axes[1][1].plot(x, y)
```

```python
# NOTE! This returns 2 dimensional array
fig, axes = plt.subplots(nrows = 2, ncols= 2)

axes[0][0].plot(a, b)
axes[1][1].plot(x, y)

plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows = 2, ncols = 2, figsize = (12, 8))
# SET YOUR AXES PARAMETERS FIRST
# Parameters at the axes level
axes[0][0].plot(a, b)
axes[0][0].set_title("Number 1 title")
axes[1][1].plot(x,y)
axes[1][1].set_title('Number 4 Title')
axes[1][1].set_xlabel('1 1 X Label')
axes[0][1].plot(y,x)
axes[1][0].plot(b,a)
# THEN SET OVERALL FIGURE PARAMETERS
fig.suptitle("Figure Level", fontsize = 16)
# Parameters at the Figure level
```

```python
fig,axes = plt.subplots(nrows=2,ncols=2,figsize=(12,8))
# Parameters at the axes level
axes[0][0].plot(a,b)
axes[1][1].plot(x,y)
axes[0][1].plot(y,x)
axes[1][0].plot(b,a)
# Use left,right,top, bottom to stretch subplots
# Use wspace,hspace to add spacing between subplots
fig.subplots_adjust(
left=None,
bottom=None,
right=None,
top=None,
wspace=0.9,
hspace=0.6)
```

```python
# NOTE! This returns 2 dimensional array
fig,axes = plt.subplots(nrows=2,ncols=2,figsize=(12,8))
# Parameters at the axes level
axes[0][0].plot(a,b)
axes[1][1].plot(x,y)
axes[0][1].plot(y,x)
axes[1][0].plot(b,a)
fig.savefig("subplots.png", bbox_inches = "tight") # save new image file 
```
