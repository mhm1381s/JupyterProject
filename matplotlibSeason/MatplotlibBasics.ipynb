```python
plt.plot(x, y)
plt.xlabel("X Axis Title Here")
plt.ylabel("Y Axis Title Here")
plt.title("String Title Here", loc = 'left') 
plt.show() # Required for non-jupyter users , but also removes Out[] info
```

```python
plt.plot(x, y)
plt.xlabel("X Axis Title Here")
plt.ylabel("Y Axis Title Here")
plt.title("String Title Here", loc = 'center') 
plt.show() # Required for non-jupyter users , but also removes Out[] info
```

```python
plt.plot(x, y)
plt.xlabel("X Axis Title Here")
plt.ylabel("Y Axis Title Here")
plt.title("String Title Here", loc = 'left')
plt.xlim(0, 10) # Lower Limit, Upper Limit
plt.ylim(0, 20) # Lower Limit, Upper Limit
plt.show() # Required for non-jupyter users , but also removes Out[] info
```
