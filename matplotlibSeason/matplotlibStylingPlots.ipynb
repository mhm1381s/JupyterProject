```python
fig, ax = plt.subplots()
ax.plot(x, x + 1, color = "blue", alpha = 0.5) # half-transparant
ax.plot(x, x + 2, color = "#A521A5") # RGB hex code
ax.plot(x, x + 3, color = "#FF8C00") # RGB hex code
```

```python
fig, ax = plt.subplots(figsize=(12,6))
# Use linewidth or lw
ax.plot(x, x - 1, color = "blue", linewidth = 0.25)
ax.plot(x, x - 2, color = "blue", lw = 0.5)
ax.plot(x, x - 3, color = "blue", lw = 1)
ax.plot(x, x - 4, color = "blue", lw = 10)
```

```python
# possible linestype options ‘--‘, ‘–’, ‘-.’, ‘:’, ‘steps’
fig, ax = plt.subplots(figsize=(12,6))
ax.plot(x, x-1, color="green", lw=3, linestyle = '-' ) # solid
ax.plot(x, x-2, color="green", lw=3, linestyle = '-.' ) # dash and dot
ax.plot(x, x-3, color="green", lw=3, linestyle = ':' ) # dots
ax.plot(x, x-4, color="green", lw=3, linestyle = '--' ) # dashes
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,1,3,5]

#plottig
plt.plot(x, y , marker= 'o', linestyle= '-', color= 'blue', label= 'Sample Data')

# Setting major ticks on x-axis and y-axis
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([1, 2, 3, 4, 5])

# Adding labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sample Plot with Major Ticks')
# Adding legend
plt.legend()
# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]
# Plotting
plt.plot(x, y, marker='o', linestyle='-', color='blue', label='Sample Data')
# Setting major ticks on x-axis and y-axis
plt.gca().xaxis.set_minor_locator(MultipleLocator(0.5))
plt.gca().yaxis.set_minor_locator(MultipleLocator(0.5))
# Adding labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sample Plot with Major Ticks')
# Adding legend
plt.legend()
# Show plot
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]
# Plotting
plt.plot(x, y, marker='o', linestyle='-', color='blue', label='Sample Data')
# Adding grid
plt.grid(True)
# Adding labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sample Plot with Grid')
plt.grid(True, linestyle='--', linewidth=0.5, color='gray', alpha=0.7)
# Adding legend
plt.legend()
# Show plot
plt.show()
```
