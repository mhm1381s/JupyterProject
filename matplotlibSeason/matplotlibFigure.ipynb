```python
# Create Figure (empty canvas)
fig = plt.figure()
# Add set of axes to figure
axes = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height (range 0 to 1)
# Plot on that set of axes
axes.plot(x, y)
plt.show()
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0, 0, 1, 1])
axes2 = fig.add_axes([0.2, 0.2, 0.5, 0.5])

axes1.plot(a, b)
axes1.set_xlabel("X label")
axes1.set_ylabel("Y label")
axes1.set_title("Big Figure")

axes2.plot(a, b)
axes2.set_title("Small Figure")
```

```python
# Creates blank canvas
fig = plt.figure()
# Larger Figure Axes 1
axes1 = fig.add_axes([0, 0, 1, 1])
axes2 = fig.add_axes([0.2, 0.5, 0.25, 0.25])
# Larger Figure Axes 1
axes1.plot(a, b)
```

```python
# Creates blank canvas
fig = plt.figure()
# Larger Figure Axes 1
axes1 = fig.add_axes([0, 0, 1, 1])
axes2 = fig.add_axes([0.2, 0.5, 0.25, 0.25])
# Larger Figure Axes 1
axes1.plot(a, b)
# Use set_ to add to the axes figure
axes1.set_xlabel("X Label")
axes1.set_ylabel("Y Label")
axes1.set_title("Big Figure")
# Insert Figure Axes 2
axes2.plot(a, b)
axes2.set_xlim(8, 10)
axes2.set_ylim(4000, 10000)
axes2.set_xlabel("X")
axes2.set_ylabel("Y")
axes2.set_title("Zoom In")
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0, 0, 1, 1])
axes2 = fig.add_axes([0.2, 0.5, 0.25, 0.25])
axes3 = fig.add_axes([1, 1, 0.25, 0.25])

axes1.plot(a, b)
axes1.set_xlabel("X Label")
axes1.set_ylabel("Y Label")
axes1.set_title("Big Figure")

axes2.plot(a, b)
axes2.set_xlim(8, 10)
axes2.set_ylim(4000, 10000)
axes2.set_xlabel("X")
axes2.set_ylabel("Y")
axes2.set_title("Zoom In")

axes3.plot(x, y)

plt.show()
```

```python
# Creates blank canvas
plt.rcParams["font.family"] = "Times New Roman"
fig = plt.figure(figsize = (12, 8), dpi = 400)
axes1 = fig.add_axes([0, 0, 1, 1])
axes1.set_xlabel('X Label', fontname = "Times New Roman", fontsize = 25)
axes1.set_ylabel('Y Label', fontname = "Times New Roman", fontsize = 25)
axes1.plot(a, b)


```

```python
# Creates blank canvas
fig = plt.figure(figsize=(12,8), dpi = 300)
axes1 = fig.add_axes([0, 0, 1, 1]) # Full figure
axes2 = fig.add_axes([1, 1, 0.25, 0.25]) # Starts at top right corner!
# Larger Figure Axes 1
axes1.plot(x,y)
# Insert Figure Axes 2
axes2.plot(x,y)
fig.savefig('test.png',bbox_inches='tight')
```
