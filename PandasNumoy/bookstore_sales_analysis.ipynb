# تحلیل فروش کتابفروشی داستانی
## با استفاده از پانداس

```python
import pandas as pd
```

```python
# ایجاد داده‌های نمونه برای فروش کتابفروشی
data = {
    'date': ['2024-01-15', '2024-01-15', '2024-01-16', '2024-01-16', '2024-01-17', '2024-01-17',
             '2024-01-18', '2024-01-18', '2024-01-19', '2024-01-19', '2024-01-20', '2024-01-20'],
    'book_title': ['گتسبی بزرگ', 'کشتن مرغ مقلد', 'هری پاتر و سنگ جادو', '1984', 'غرور و تعصب', 'هابیت',
                   'ناتور دشت', 'مزرعه حیوانات', 'ارباب حلقه‌ها', 'دنیای قشنگ نو', 'نارنیا', 'فرانکنشتاین'],
    'genre': ['کلاسیک', 'کلاسیک', 'فانتزی', 'دیستوپی', 'کلاسیک', 'فانتزی',
              'کلاسیک', 'دیستوپی', 'فانتزی', 'دیستوپی', 'فانتزی', 'وحشت'],
    'author': ['اف. اسکات فیتزجرالد', 'هارپر لی', 'جی.کی. رولینگ', 'جرج اورول', 'جین آستن', 'جی.آر.آر. تالکین',
               'جی.دی. سلینجر', 'جرج اورول', 'جی.آر.آر. تالکین', 'آلدوس هاکسلی', 'سی.اس. لوئیس', 'مری شلی'],
    'price': [129000, 109000, 149000, 99000, 119000, 139000,
              109000, 89000, 169000, 99000, 129000, 109000],
    'quantity_sold': [5, 3, 8, 4, 6, 7, 2, 5, 9, 3, 6, 4],
    'store_location': ['تهران', 'تهران', 'اصفهان', 'اصفهان', 'مشهد', 'مشهد',
                       'تهران', 'اصفهان', 'مشهد', 'تهران', 'اصفهان', 'مشهد']
}

df = pd.DataFrame(data)
df
```

```python
# اطلاعات پایه
print("اطلاعات پایه داده‌ها:")
print(f"تعداد کل رکوردها: {len(df)}")
print(f"تاریخ شروع: {df['date'].min()}")
print(f"تاریخ پایان: {df['date'].max()}")
print(f"تعداد کتاب‌های فروخته شده: {df['quantity_sold'].sum()}")
print(f"درآمد کل: {df['price'].dot(df['quantity_sold']):,} تومان")
```

```python
# فروش بر اساس ژانر
genre_sales = df.groupby('genre')['quantity_sold'].sum().sort_values(ascending=False)
print("فروش بر اساس ژانر:")
print(genre_sales)
```

```python
# درآمد بر اساس ژانر
df['revenue'] = df['price'] * df['quantity_sold']
genre_revenue = df.groupby('genre')['revenue'].sum().sort_values(ascending=False)
print("درآمد بر اساس ژانر:")
print(genre_revenue.apply(lambda x: f"{x:,} تومان"))
```

```python
# پرفروش‌ترین کتاب‌ها
top_books = df.groupby('book_title')['quantity_sold'].sum().sort_values(ascending=False)
print("پرفروش‌ترین کتاب‌ها:")
print(top_books)
```

```python
# فروش بر اساس شهر
store_sales = df.groupby('store_location')['quantity_sold'].sum().sort_values(ascending=False)
print("فروش بر اساس شهر:")
print(store_sales)
```

```python
# پرفروش‌ترین نویسندگان
author_sales = df.groupby('author')['quantity_sold'].sum().sort_values(ascending=False)
print("پرفروش‌ترین نویسندگان:")
print(author_sales)
```

```python
# فروش روزانه
daily_sales = df.groupby('date')['quantity_sold'].sum()
print("فروش روزانه:")
print(daily_sales)
```

```python
# آمارهای خلاصه
print("آمارهای خلاصه:")
print(f"میانگین قیمت کتاب: {df['price'].mean():,.0f} تومان")
print(f"میانگین تعداد فروش در هر تراکنش: {df['quantity_sold'].mean():.1f}")
print(f"ژانر محبوب: {genre_sales.idxmax()} ({genre_sales.max()} کتاب)")
print(f"شهر با بیشترین فروش: {store_sales.idxmax()} ({store_sales.max()} کتاب)")
print(f"نویسنده پرفروش: {author_sales.idxmax()} ({author_sales.max()} کتاب)")
```

```python
# عملیات‌های مختلف روی داده‌ها
print("\nعملیات‌های مختلف روی داده‌ها:")

# فیلتر کردن داده‌ها
fantasy_books = df[df['genre'] == 'فانتزی']
print(f"تعداد کتاب‌های فانتزی: {len(fantasy_books)}")

# انتخاب ستون‌های خاص
selected_columns = df[['book_title', 'author', 'price', 'quantity_sold']]
print("\nستون‌های انتخاب شده:")
print(selected_columns.head())

# مرتب‌سازی
sorted_by_price = df.sort_values('price', ascending=False)
print("\nمرتب شده بر اساس قیمت (گران‌ترین):")
print(sorted_by_price[['book_title', 'price']].head())

# محاسبه ستون جدید
df['total_sale'] = df['price'] * df['quantity_sold']
print("\nداده‌ها با ستون total_sale:")
print(df[['book_title', 'price', 'quantity_sold', 'total_sale']].head())
```
